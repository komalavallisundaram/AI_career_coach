```python
# =====================================================
# IMPORTS
# =====================================================

import pandas as pd
import pickle

from sklearn.feature_extraction.text import TfidfVectorizer

# =====================================================
# FILE PATHS
# =====================================================

DATA_PATH = (
    r"C:\Users\shenbagam\videos\history\master_dataset_cleaned.csv"
)

TFIDF_PKL = (
    r"C:\Users\shenbagam\videos\history\tfidf_matrix.pkl"
)

VECTORIZER_PKL = (
    r"C:\Users\shenbagam\videos\history\tfidf_vectorizer.pkl"
)

# =====================================================
# LOAD DATASET
# =====================================================

print("Loading Dataset...")

df = pd.read_csv(
    DATA_PATH,
    encoding="utf-8-sig"
)

print(
    "Dataset Loaded Successfully"
)

print(
    "Shape:",
    df.shape
)

# =====================================================
# VERIFY COLUMN
# =====================================================

if "embedding_text" not in df.columns:

    raise ValueError(
        "'embedding_text' column not found."
    )

# =====================================================
# CLEAN NULL VALUES
# =====================================================

df["embedding_text"] = (

    df["embedding_text"]

    .fillna("")

    .astype(str)
)

# =====================================================
# CREATE TF-IDF VECTORIZER
# =====================================================

print(
    "\nCreating TF-IDF Matrix..."
)

vectorizer = TfidfVectorizer(

    stop_words="english",

    lowercase=True,

    max_features=10000
)

tfidf_matrix = vectorizer.fit_transform(

    df["embedding_text"]
)

print(
    "TF-IDF Matrix Created"
)

print(
    "TF-IDF Shape:",
    tfidf_matrix.shape
)

# =====================================================
# SAVE TF-IDF MATRIX
# =====================================================

with open(
    TFIDF_PKL,
    "wb"
) as f:

    pickle.dump(
        tfidf_matrix,
        f
    )

# =====================================================
# SAVE VECTORIZER
# =====================================================

with open(
    VECTORIZER_PKL,
    "wb"
) as f:

    pickle.dump(
        vectorizer,
        f
    )

# =====================================================
# SUCCESS MESSAGE
# =====================================================

print("\n" + "=" * 50)

print(
    "TF-IDF Matrix Saved Successfully"
)

print(
    TFIDF_PKL
)

print(
    "\nTF-IDF Vectorizer Saved Successfully"
)

print(
    VECTORIZER_PKL
)

print("=" * 50)
```
